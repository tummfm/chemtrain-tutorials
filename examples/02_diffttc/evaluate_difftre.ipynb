# Evaluate the DiffTRe potential

```python
from pathlib import Path
import shlex
import subprocess
import sys

TUTORIAL = Path.cwd().resolve()
if not (TUTORIAL / "evaluate_pinning.py").is_file():
    TUTORIAL = TUTORIAL / "examples" / "02_diffttc"
if not (TUTORIAL / "evaluate_pinning.py").is_file():
    raise FileNotFoundError("Could not locate examples/02_diffttc/evaluate_pinning.py")

GPU = "0"
OVERWRITE = False
MODEL = TUTORIAL / "training/output/mace_difftre.ptb"
OUTPUT = TUTORIAL / "training/output/final"

if not MODEL.is_file():
    raise FileNotFoundError(f"DiffTRe model has not been exported: {MODEL}")
```

## Interface-pinning simulation

```python
command = [sys.executable, "evaluate_pinning.py", "--gpu", GPU]
if OVERWRITE:
    command.append("--overwrite")

print(shlex.join(command))
subprocess.run(command, cwd=TUTORIAL, check=True)
```

## Interface-pinning result

```python
from IPython.display import display

from ip_helpers import (
    lammps_atom_count,
    load_pinning_parameters,
    load_runs,
    pinning_free_energy_difference,
    plot_evaluation,
    summary,
)

pinning = load_pinning_parameters(TUTORIAL / "input/interface_pinning.toml")
equilibration, production = load_runs(OUTPUT)
free_energy = pinning_free_energy_difference(
    pinning, production, lammps_atom_count(TUTORIAL / "input/interface_start.lmpdat")
)
production_summary = summary(production)
figure = plot_evaluation(
    equilibration, production, pinning, timestep_ps=0.004
)
```
